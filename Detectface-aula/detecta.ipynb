```python
import cv2
from skimage import io, color
from IPython.display import Image
```

```python
Image(url="img/grupo_pessoas.jpg")
```

```python
img_grupo1 = io.imread("img/grupo_pessoas.jpg")
io.imshow(img_grupo1)
io.show()
```

```python
img_grupo1 = cv2.imread("img/grupo_pessoas.jpg")
img_grupo1.shape
```

```python
img_grey = color.rgb2gray(img_grupo1)
io.imshow(img_grey)
io.show()
```

```python
grayImage1 = cv2.cvtColor(img_grupo1, cv2.COLOR_BGR2GRAY)
grayImage1.shape
```

```python
face_parametro = "cascade/frontalface_alt.xml"
classificador_cascade = cv2.CascadeClassifier(face_parametro)
```

```python
classificador_cascade
```

```python
faces = classificador_cascade.detectMultiScale(grayImage1, scaleFactor=1.1, minNeighbors=6, minSize=(20,20))
```

```python
faces
```

```python
for (x,y, width, height) in faces:
    cv2.rectangle(img_grupo1, (x,y), (x + width, y + height), (0, 255, 0), 6)
```

```python
cv2.imwrite("img/resultado_grupo1.jpg", img_grupo1)
```
